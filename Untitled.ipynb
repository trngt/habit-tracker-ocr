```python
#!/usr/bin/env python3
"""
Habit Tracker Scanner - Steps 1-3
Detects corner calibration markers and applies perspective correction
"""

import cv2
import numpy as np
from typing import List, Tuple, Optional
import sys


def load_and_preprocess(image_path: str) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Step 1: Load image and create preprocessed versions
    
    Args:
        image_path: Path to the scanned habit tracker image
        
    Returns:
        Tuple of (original_color, grayscale, binary_threshold)
    """
    # Load the image
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"Could not load image at {image_path}")
    
    print(f"Loaded image: {image.shape[1]}x{image.shape[0]} pixels")
    
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Apply binary threshold (invert so markers are white on black background)
    _, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)
    
    return image, gray, binary


def detect_corner_markers(binary: np.ndarray, 
                          min_area: int = 500,
                          max_area: int = 10000) -> List[Tuple[int, int]]:
    """
    Step 2: Detect the 4 corner calibration markers
    
    Args:
        binary: Binary thresholded image
        min_area: Minimum contour area to consider as marker
        max_area: Maximum contour area to consider as marker
        
    Returns:
        List of 4 (x, y) corner coordinates sorted as:
        [top-left, top-right, bottom-left, bottom-right]
    """
    # Find all contours
    contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    print(f"Found {len(contours)} contours")
    
    # Filter for marker candidates
    markers = []
    for cnt in contours:
        area = cv2.contourArea(cnt)
        
        # Filter by area
        if min_area < area < max_area:
            # Get bounding rectangle
            x, y, w, h = cv2.boundingRect(cnt)
            
            # Check if roughly square (markers should be square)
            aspect_ratio = w / h if h > 0 else 0
            if 0.7 < aspect_ratio < 1.3:  # Allow some tolerance
                # Use center of bounding box as marker position
                center_x = x + w // 2
                center_y = y + h // 2
                markers.append((center_x, center_y, area))
                print(f"  Marker candidate at ({center_x}, {center_y}), area={area:.0f}")
    
    #if len(markers) < 4:
    #    raise ValueError(f"Found only {len(markers)} markers, need 4 corner markers")
    
    if len(markers) > 4:
        # Take the 4 largest markers (most likely to be our calibration squares)
        markers.sort(key=lambda m: m[2], reverse=True)
        markers = markers[:4]
        print(f"Found {len(markers)} candidates, using 4 largest")
    
    # Remove area from tuples, keep just (x, y)
    marker_points = [(x, y) for x, y, _ in markers]
    
    # Sort markers into consistent order: TL, TR, BL, BR
    sorted_markers = sort_corners(marker_points)
    
    return sorted_markers


def sort_corners(points: List[Tuple[int, int]]) -> List[Tuple[int, int]]:
    """
    Sort 4 corner points into consistent order:
    [top-left, top-right, bottom-left, bottom-right]
    
    Args:
        points: List of 4 (x, y) coordinates
        
    Returns:
        Sorted list of 4 coordinates
    """
    # Convert to numpy array for easier manipulation
    pts = np.array(points, dtype=np.float32)
    
    # Sort by y-coordinate to get top 2 and bottom 2
    sorted_by_y = pts[np.argsort(pts[:, 1])]
    top_two = sorted_by_y[:2]
    bottom_two = sorted_by_y[2:]
    
    # Sort each pair by x-coordinate
    top_two = top_two[np.argsort(top_two[:, 0])]  # left to right
    bottom_two = bottom_two[np.argsort(bottom_two[:, 0])]  # left to right
    
    # Return in order: TL, TR, BL, BR
    sorted_pts = [
        tuple(top_two[0]),      # top-left
        tuple(top_two[1]),      # top-right
        tuple(bottom_two[0]),   # bottom-left
        tuple(bottom_two[1])    # bottom-right
    ]
    
    print("\nSorted corners:")
    print(f"  Top-left:     {sorted_pts[0]}")
    print(f"  Top-right:    {sorted_pts[1]}")
    print(f"  Bottom-left:  {sorted_pts[2]}")
    print(f"  Bottom-right: {sorted_pts[3]}")
    
    return sorted_pts


def apply_perspective_correction(image: np.ndarray,
                                 corners: List[Tuple[int, int]],
                                 output_width: int = 1700,
                                 output_height: int = 2200) -> np.ndarray:
    """
    Step 3: Apply perspective transform to straighten the image
    
    Args:
        image: Input image (color or grayscale)
        corners: List of 4 corner coordinates [TL, TR, BL, BR]
        output_width: Desired width of output image
        output_height: Desired height of output image
        
    Returns:
        Warped/corrected image
    """
    # Source points (detected corners)
    src_points = np.array(corners, dtype=np.float32)
    
    # Destination points (perfect rectangle)
    dst_points = np.array([
        [0, 0],                           # top-left
        [output_width - 1, 0],           # top-right
        [0, output_height - 1],          # bottom-left
        [output_width - 1, output_height - 1]  # bottom-right
    ], dtype=np.float32)
    
    # Calculate perspective transform matrix
    matrix = cv2.getPerspectiveTransform(src_points, dst_points)
    
    # Apply the transform
    warped = cv2.warpPerspective(image, matrix, (output_width, output_height))
    
    print(f"\nPerspective correction applied")
    print(f"Output size: {output_width}x{output_height}")
    
    return warped


def visualize_detection(image: np.ndarray, 
                       corners: List[Tuple[int, int]], 
                       output_path: str = "detected_markers.jpg"):
    """
    Helper function to visualize detected markers (for debugging)
    
    Args:
        image: Original image
        corners: Detected corner coordinates
        output_path: Where to save visualization
    """
    vis = image.copy()
    
    # Draw circles at detected corners
    colors = [(0, 0, 255), (0, 255, 0), (255, 0, 0), (255, 255, 0)]  # BGR
    labels = ['TL', 'TR', 'BL', 'BR']
    
    for i, (x, y) in enumerate(corners):
        cv2.circle(vis, (int(x), int(y)), 20, colors[i], -1)
        cv2.putText(vis, labels[i], (int(x) - 30, int(y) - 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, colors[i], 2)
    
    # Draw lines connecting corners to show detected quadrilateral
    pts = np.array(corners, dtype=np.int32)
    pts = pts.reshape((-1, 1, 2))
    cv2.polylines(vis, [pts[[0, 1, 3, 2]]], True, (0, 255, 255), 3)
    
    cv2.imwrite(output_path, vis)
    print(f"Visualization saved to {output_path}")

```

```python

def scan_image(image_path: str, output_dir: str):
    """
    Main pipeline for steps 1-3
    """
    print("=" * 60)
    print("HABIT TRACKER SCANNER - Steps 1-3")
    print("=" * 60)
    
    # Step 1: Load and preprocess
    print("\n[Step 1] Loading and preprocessing image...")
    original, gray, binary = load_and_preprocess(image_path)
    
    # Step 2: Detect corner markers
    print("\n[Step 2] Detecting corner markers...")
    corners = detect_corner_markers(binary)
    
    # Visualize detection (optional, for debugging)
    output_path = f"{output_dir}/detected_markers.jpg"
    visualize_detection(original, corners, output_path)
    
    # Step 3: Apply perspective correction
    print("\n[Step 3] Applying perspective correction...")
    warped = apply_perspective_correction(original, corners)
    
    # Save the corrected image
    output_path = f"{output_dir}/corrected_grid.jpg"
    cv2.imwrite(output_path, warped)
    print(f"\nCorrected image saved to {output_path}")
    
    # Also save grayscale version for further processing
    warped_gray = cv2.cvtColor(warped, cv2.COLOR_BGR2GRAY)
    cv2.imwrite(f"{output_dir}/corrected_grid_gray.jpg", warped_gray)
    print(f"Grayscale version saved to corrected_grid_gray.jpg")
    
    print("\n" + "=" * 60)
    print("SUCCESS! Steps 1-3 complete.")
    print("=" * 60)
    
    return warped, corners
```

```python
#!/usr/bin/env python3
"""
Habit Tracker Scanner - Steps 4-5
Calculate grid cell positions and detect filled bubbles
"""

import cv2
import numpy as np
from typing import List, Tuple, Dict
import sys
import json


class GridConfig:
    """Configuration for the habit tracker grid"""
    def __init__(self):
        # Grid structure
        self.num_rows = 31  # Days 1-31
        self.num_cols = 16  # Number of habit columns
        
        # Grid position in the corrected image (adjust based on your layout)
        # These are offsets from the edges after perspective correction
        self.grid_left = 150      # pixels from left edge
        self.grid_top = 400       # pixels from top edge
        self.grid_width = 1000    # total grid width in pixels
        self.grid_height = 1650   # total grid height in pixels
        
        # Bubble detection parameters
        self.fill_threshold = 0.15  # If >15% of cell pixels are dark, consider filled
        self.cell_padding = 0.2     # Sample inner 80% of cell (avoid borders)


def calculate_grid_cells(config: GridConfig) -> List[List[Tuple[int, int, int, int]]]:
    """
    Step 4: Calculate pixel coordinates for each grid cell
    
    Args:
        config: GridConfig with grid dimensions and position
        
    Returns:
        2D list [row][col] of (x, y, width, height) for each cell
    """
    print(f"\n[Step 4] Calculating grid cell positions...")
    print(f"  Grid: {config.num_rows} rows x {config.num_cols} cols")
    print(f"  Position: ({config.grid_left}, {config.grid_top})")
    print(f"  Size: {config.grid_width} x {config.grid_height}")
    
    cell_width = config.grid_width / config.num_cols
    cell_height = config.grid_height / config.num_rows
    
    print(f"  Cell size: {cell_width:.1f} x {cell_height:.1f} pixels")
    
    grid_cells = []
    for row in range(config.num_rows):
        row_cells = []
        for col in range(config.num_cols):
            x = int(config.grid_left + col * cell_width)
            y = int(config.grid_top + row * cell_height)
            w = int(cell_width)
            h = int(cell_height)
            row_cells.append((x, y, w, h))
        grid_cells.append(row_cells)
    
    print(f"  Total cells: {config.num_rows * config.num_cols}")
    return grid_cells


def detect_filled_bubble(image_gray: np.ndarray, 
                        cell_rect: Tuple[int, int, int, int],
                        config: GridConfig) -> Tuple[bool, float]:
    """
    Detect if a bubble/cell is filled in
    
    Args:
        image_gray: Grayscale image
        cell_rect: (x, y, width, height) of the cell
        config: GridConfig with detection parameters
        
    Returns:
        Tuple of (is_filled: bool, fill_ratio: float)
    """
    x, y, w, h = cell_rect
    
    # Apply padding to sample inner region (avoid grid lines)
    pad_x = int(w * config.cell_padding)
    pad_y = int(h * config.cell_padding)
    
    # Extract the inner region of the cell
    roi = image_gray[
        y + pad_y : y + h - pad_y,
        x + pad_x : x + w - pad_x
    ]
    
    if roi.size == 0:
        return False, 0.0
    
    # Threshold to binary (dark pixels = filled marks)
    # Lower values = darker pixels
    _, binary = cv2.threshold(roi, 180, 255, cv2.THRESH_BINARY_INV)
    
    # Calculate fill ratio (percentage of dark pixels)
    dark_pixels = np.sum(binary == 255)
    total_pixels = binary.size
    fill_ratio = dark_pixels / total_pixels if total_pixels > 0 else 0.0
    
    # Determine if filled based on threshold
    is_filled = fill_ratio > config.fill_threshold
    
    return is_filled, fill_ratio


def detect_all_bubbles(image_gray: np.ndarray,
                      grid_cells: List[List[Tuple[int, int, int, int]]],
                      config: GridConfig) -> Dict[int, List[int]]:
    """
    Step 5: Detect all filled bubbles in the grid
    
    Args:
        image_gray: Grayscale corrected image
        grid_cells: 2D list of cell rectangles [row][col]
        config: GridConfig with detection parameters
        
    Returns:
        Dictionary mapping day (1-31) to list of filled habit column indices
    """
    print(f"\n[Step 5] Detecting filled bubbles...")
    print(f"  Fill threshold: {config.fill_threshold * 100:.1f}%")
    
    results = {}
    total_filled = 0
    
    for row in range(config.num_rows):
        day = row + 1  # Days are 1-indexed
        filled_habits = []
        
        for col in range(config.num_cols):
            cell_rect = grid_cells[row][col]
            is_filled, fill_ratio = detect_filled_bubble(image_gray, cell_rect, config)
            
            if is_filled:
                filled_habits.append(col)
                total_filled += 1
        
        if filled_habits:
            results[day] = filled_habits
    
    print(f"  Total filled bubbles detected: {total_filled}")
    print(f"  Days with activity: {len(results)}")
    
    return results


def visualize_detection(image: np.ndarray,
                       grid_cells: List[List[Tuple[int, int, int, int]]],
                       detection_results: Dict[int, List[int]],
                       config: GridConfig,
                       output_path: str = "bubble_detection.jpg"):
    """
    Create visualization showing detected filled bubbles
    
    Args:
        image: Original corrected image
        grid_cells: 2D list of cell rectangles
        detection_results: Dictionary of day -> filled habit columns
        config: GridConfig
        output_path: Where to save visualization
    """
    print(f"\n[Visualization] Creating annotated image...")
    
    vis = image.copy()
    
    # Draw all grid cells (light gray)
    for row in range(config.num_rows):
        for col in range(config.num_cols):
            x, y, w, h = grid_cells[row][col]
            cv2.rectangle(vis, (x, y), (x + w, y + h), (200, 200, 200), 1)
    
    # Highlight detected filled cells
    for day, habit_cols in detection_results.items():
        row = day - 1  # Convert back to 0-indexed
        for col in habit_cols:
            x, y, w, h = grid_cells[row][col]
            
            # Draw green rectangle around detected filled cell
            cv2.rectangle(vis, (x, y), (x + w, y + h), (0, 255, 0), 3)
            
            # Add checkmark
            center_x = x + w // 2
            center_y = y + h // 2
            cv2.drawMarker(vis, (center_x, center_y), (0, 200, 0),
                          cv2.MARKER_TILTED_CROSS, 15, 2)
    
    # Add legend
    legend_y = 50
    cv2.putText(vis, "Green = Detected Filled", (50, legend_y),
               cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.putText(vis, f"Total: {sum(len(cols) for cols in detection_results.values())} bubbles",
               (50, legend_y + 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)
    
    cv2.imwrite(output_path, vis)
    print(f"  Visualization saved to {output_path}")


def visualize_fill_ratios(image_gray: np.ndarray,
                         grid_cells: List[List[Tuple[int, int, int, int]]],
                         config: GridConfig,
                         output_path: str = "fill_ratios_heatmap.jpg"):
    """
    Create heatmap showing fill ratios for all cells (debugging aid)
    
    Args:
        image_gray: Grayscale corrected image
        grid_cells: 2D list of cell rectangles
        config: GridConfig
        output_path: Where to save heatmap
    """
    print(f"\n[Debug Visualization] Creating fill ratio heatmap...")
    
    # Create a blank image for the heatmap
    h = config.num_rows * 30
    w = config.num_cols * 30
    heatmap = np.ones((h, w, 3), dtype=np.uint8) * 255
    
    cell_h = h // config.num_rows
    cell_w = w // config.num_cols
    
    for row in range(config.num_rows):
        for col in range(config.num_cols):
            cell_rect = grid_cells[row][col]
            _, fill_ratio = detect_filled_bubble(image_gray, cell_rect, config)
            
            # Map fill ratio to color (blue = empty, red = filled)
            # Using HSV color space for better visualization
            hue = int((1 - fill_ratio) * 120)  # 120 = blue, 0 = red
            sat = 255
            val = 255
            color_hsv = np.uint8([[[hue, sat, val]]])
            color_bgr = cv2.cvtColor(color_hsv, cv2.COLOR_HSV2BGR)[0][0]
            
            # Draw cell
            y1 = row * cell_h
            x1 = col * cell_w
            cv2.rectangle(heatmap, (x1, y1), (x1 + cell_w, y1 + cell_h),
                         color_bgr.tolist(), -1)
            
            # Add text showing ratio
            text = f"{fill_ratio:.2f}"
            font_scale = 0.3
            cv2.putText(heatmap, text, (x1 + 2, y1 + cell_h - 5),
                       cv2.FONT_HERSHEY_SIMPLEX, font_scale, (0, 0, 0), 1)
    
    cv2.imwrite(output_path, heatmap)
    print(f"  Heatmap saved to {output_path}")


def export_results_json(detection_results: Dict[int, List[int]],
                       output_path: str = "detection_results.json"):
    """
    Export detection results to JSON format
    
    Args:
        detection_results: Dictionary of day -> filled habit columns
        output_path: Where to save JSON file
    """
    print(f"\n[Export] Saving results to JSON...")
    
    # Convert to more readable format
    formatted_results = {
        "detected_bubbles": {
            f"day_{day}": {
                "day_number": day,
                "filled_habits": habit_cols
            }
            for day, habit_cols in detection_results.items()
        },
        "summary": {
            "total_days_with_activity": len(detection_results),
            "total_filled_bubbles": sum(len(cols) for cols in detection_results.values())
        }
    }
    
    with open(output_path, 'w') as f:
        json.dump(formatted_results, f, indent=2)
    
    print(f"  Results saved to {output_path}")


def print_results_summary(detection_results: Dict[int, List[int]]):
    """
    Print human-readable summary of detection results
    
    Args:
        detection_results: Dictionary of day -> filled habit columns
    """
    print("\n" + "=" * 60)
    print("DETECTION RESULTS SUMMARY")
    print("=" * 60)
    
    if not detection_results:
        print("No filled bubbles detected.")
        return
    
    print(f"\nTotal days with activity: {len(detection_results)}")
    print(f"Total filled bubbles: {sum(len(cols) for cols in detection_results.values())}")
    
    print("\nDetailed results:")
    for day in sorted(detection_results.keys()):
        habit_cols = detection_results[day]
        # Convert to 1-indexed for display
        habit_numbers = [col + 1 for col in habit_cols]
        print(f"  Day {day:2d}: Habits {habit_numbers}")
    
    print("=" * 60)

```

```python

def main(corrected_image_path: str):
    """
    Main pipeline for steps 4-5
    
    Args:
        corrected_image_path: Path to the perspective-corrected image
    """
    print("=" * 60)
    print("HABIT TRACKER SCANNER - Steps 4-5")
    print("=" * 60)
    
    # Load the corrected image
    print(f"\nLoading corrected image: {corrected_image_path}")
    image = cv2.imread(corrected_image_path)
    if image is None:
        raise ValueError(f"Could not load image at {corrected_image_path}")
    
    image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    print(f"Image size: {image.shape[1]}x{image.shape[0]}")
    
    # Initialize configuration
    config = GridConfig()
    
    # Step 4: Calculate grid cells
    grid_cells = calculate_grid_cells(config)
    
    # Step 5: Detect filled bubbles
    detection_results = detect_all_bubbles(image_gray, grid_cells, config)
    
    # Create visualizations
    visualize_detection(image, grid_cells, detection_results, config)
    visualize_fill_ratios(image_gray, grid_cells, config)
    
    # Export results
    export_results_json(detection_results)
    
    # Print summary
    print_results_summary(detection_results)
    
    print("\n" + "=" * 60)
    print("SUCCESS! Steps 4-5 complete.")
    print("=" * 60)
    print("\nGenerated files:")
    print("  - bubble_detection.jpg      (annotated image with detected fills)")
    print("  - fill_ratios_heatmap.jpg   (heatmap showing fill ratios)")
    print("  - detection_results.json    (machine-readable results)")
    
    return detection_results

main('/Users/trung/Projects/Habits/output/corrected_grid_gray.jpg')
```
